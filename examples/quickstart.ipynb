# IDSS one-call quickstart
Synthetic demonstration data only. Install from the repository root with `python -m pip install .`. The path setup below also supports running directly from this source tree.


```python
from pathlib import Path
import sys
root = Path.cwd()
if not (root / "src" / "idss").exists():
    root = root.parent
assert (root / "src" / "idss").exists(), "Open in the repository root or examples folder"
sys.path.insert(0, str(root / "src"))
from idss import IDSS
csv_path = root / "examples" / "models.csv"
result = IDSS(csv_path)
print(result)
print(result.best)
```

## Custom weights
Tuple order is quality, latency, size. The dictionary form names each criterion.


```python
custom = IDSS(csv_path, weights=(0.5, 0.3, 0.2))
print(custom)
named = IDSS(csv_path, weights={"quality": 0.5, "latency": 0.3, "size": 0.2})
assert custom.rankings == named.rankings
```

## Standard profiles


```python
print(IDSS.profiles())
standard = IDSS(csv_path, profile="quality_priority")
print(standard)
```

## Rankings, sensitivity and export


```python
rankings = result.rankings
sensitivity = result.sensitivity()
output = result.save(root / "idss_results")
print(output / "report.html")
```
